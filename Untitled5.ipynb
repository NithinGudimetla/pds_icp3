```python
import pandas as pd

# ✅ Your provided working URL
url = "https://raw.githubusercontent.com/JuliaData/CSV.jl/refs/heads/main/test/testfiles/Sacramentorealestatetransactions.csv"
df = pd.read_csv(url)

# 1. Number of columns in the data frame
print("Number of columns:", df.shape[1])

# 2. Mean of house prices
print("Mean house price:", df['price'].mean())

# 3. Median of house prices
print("Median house price:", df['price'].median())

# 4. Different types of property
print("Property types:", df['type'].unique())

# 5. Maximum price of a 'Condo'
max_condo_price = df[df['type'] == 'Condo']['price'].max()
print("Max price for Condo:", max_condo_price)

# 6. Mean price of each property type
print("Mean price by property type:\n", df.groupby('type')['price'].mean())

# 7. Count of each property type
print("Count of each property type:\n", df['type'].value_counts())
```
